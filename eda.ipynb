```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
tips = sns.load_dataset('tips')
```

```python
plt.figure(figsize=(8, 6))
sns.histplot(data=tips, x='total_bill', kde=True, bins=20)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill ($)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Total Bill Distribution by Day')
plt.xlabel('Day of Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.violinplot(data=tips, x='day', y='tip')
plt.title('Tip Distribution by Day')
plt.xlabel('Day of Week')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.barplot(data=tips, x='day', y='total_bill', estimator=np.mean)
plt.title('Average Total Bill by Day')
plt.xlabel('Day of Week')
plt.ylabel('Average Total Bill ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=tips, x='day')
plt.title('Number of Customers by Day')
plt.xlabel('Day of Week')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.title('Relationship between Total Bill and Tip by Gender')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='time', style='smoker')
plt.title('Total Bill vs Tip by Time and Smoking Status')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.regplot(data=tips, x='total_bill', y='tip')
plt.title('Regression Plot: Total Bill vs Tip')
plt.xlabel('Total Bill ($)')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(data=tips, x='day', hue='time')
plt.title('Number of Customers by Day and Time')
plt.xlabel('Day of Week')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(data=tips, x='smoker', y='tip', hue='sex')
plt.title('Tip Distribution by Smoking Status and Gender')
plt.xlabel('Smoker Status')
plt.ylabel('Tip ($)')
plt.show()
```

```python
plt.figure(figsize=(8, 6))
# Select only numerical columns
numerical_cols = tips.select_dtypes(include=[np.number])
correlation_matrix = numerical_cols.corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, fmt='.2f')
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```

```python
sns.pairplot(data=tips, vars=['total_bill', 'tip', 'size'], hue='sex')
plt.suptitle('Pair Plot: Total Bill, Tip, and Size by Gender', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex')
plt.title('Total Bill by Day and Gender')
plt.xlabel('Day of Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

```python
day_variation = tips.groupby('day')['total_bill'].std().sort_values(ascending=False)
print("Standard deviation of total_bill by day:")
print(day_variation)
print(f"\nDay with highest variation in bills: {day_variation.index[0]}")
```
